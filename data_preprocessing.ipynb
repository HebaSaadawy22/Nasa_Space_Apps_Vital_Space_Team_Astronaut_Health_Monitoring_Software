```python

# Step 2: Data Quality Check

print("=== Missing Values ===")
print(df.isnull().sum())

print("\n=== Duplicate Rows ===")
print("Duplicates:", df.duplicated().sum())

print("\n=== Duplicate Reading IDs ===")
print("Duplicate IDs:", df["reading_id"].duplicated().sum())

print("\n=== Astronauts ===")
print("Number of astronauts:", df["astronaut_id"].nunique())

print("\n=== Mission Days ===")
print("Minimum day:", df["mission_day"].min())
print("Maximum day:", df["mission_day"].max())

print("\n=== Records per Astronaut ===")
print(df.groupby("astronaut_id").size())
```

```python

# Step 3: Data Validation

import pandas as pd

# Convert timestamp to datetime
df["timestamp"] = pd.to_datetime(
    df["timestamp"],
    utc=True,
    errors="coerce"
)

# Validation ranges (technical plausibility,
# NOT medical alert thresholds)
ranges = {
    "heart_rate": (20, 250),
    "systolic_bp": (50, 260),
    "diastolic_bp": (20, 180),
    "blood_glucose": (20, 600),
    "body_temperature": (30, 43),
    "sleep_hours": (0, 24),
    "mood_score": (1, 5),
    "activity_minutes": (0, 1440),
    "spo2": (0, 100),
    "cabin_co2_ppm": (0, 100000),
    "cabin_temp_c": (-20, 60),
    "cabin_humidity_pct": (0, 100),
    "radiation_dose_rate_uSv_h": (0, float("inf"))
}

print("=== Out-of-Range Values ===")

for col, (min_val, max_val) in ranges.items():
    invalid = (~df[col].between(min_val, max_val)).sum()
    print(f"{col}: {invalid}")

print("\n=== Blood Pressure Consistency ===")
invalid_bp = (
    df["systolic_bp"] <= df["diastolic_bp"]
).sum()
print("Invalid BP pairs:", invalid_bp)

print("\n=== Invalid Timestamps ===")
print(df["timestamp"].isna().sum())

print("\n=== Duplicate Astronaut-Day Records ===")
print(
    df.duplicated(
        subset=["astronaut_id", "mission_day"]
    ).sum()
)

print("\n=== Invalid Mission Days ===")
print((df["mission_day"] < 1).sum())
```

```python

# Step 4.1: Descriptive Statistics

health_columns = [
    "heart_rate",
    "systolic_bp",
    "diastolic_bp",
    "blood_glucose",
    "body_temperature",
    "sleep_hours",
    "mood_score",
    "activity_minutes",
    "spo2"
]

print("=== Health Indicators Statistics ===")

display(
    df[health_columns].describe().T.round(2)
)

print("\n=== Average Indicators per Astronaut ===")

display(
    df.groupby("astronaut_id")[health_columns]
      .mean()
      .round(2)
)
```

```python

# Step 5: Feature Engineering

df = df.sort_values(
    ["astronaut_id", "timestamp"]
).copy()

grouped = df.groupby("astronaut_id")

# Rolling baseline from previous 3 readings
features = {
    "heart_rate": "hr",
    "sleep_hours": "sleep",
    "systolic_bp": "systolic",
    "mood_score": "mood"
}

for col, prefix in features.items():
    df[f"{prefix}_rolling_3d"] = (
        grouped[col].transform(
            lambda x: x.shift(1).rolling(
                window=3,
                min_periods=3
            ).mean()
        )
    )

# Differences from previous reading
df["hr_change"] = grouped["heart_rate"].diff()
df["sleep_change"] = grouped["sleep_hours"].diff()

new_features = [
    "hr_rolling_3d",
    "sleep_rolling_3d",
    "systolic_rolling_3d",
    "mood_rolling_3d",
    "hr_change",
    "sleep_change"
]

display(
    df[
        ["astronaut_id", "mission_day"] +
        new_features
    ].head(10)
)
```

```python

# Step 5.2: Personal Baseline Deviations

df["hr_deviation"] = (
    df["heart_rate"] - df["hr_rolling_3d"]
)

df["sleep_deviation"] = (
    df["sleep_hours"] - df["sleep_rolling_3d"]
)

df["systolic_deviation"] = (
    df["systolic_bp"] - df["systolic_rolling_3d"]
)

df["mood_deviation"] = (
    df["mood_score"] - df["mood_rolling_3d"]
)

deviation_columns = [
    "astronaut_id",
    "mission_day",
    "heart_rate",
    "hr_rolling_3d",
    "hr_deviation",
    "sleep_hours",
    "sleep_rolling_3d",
    "sleep_deviation",
    "systolic_deviation",
    "mood_deviation"
]

display(
    df[deviation_columns]
    .round(2)
    .head(10)
)
```

```python

# Step 6.1 — ML Dataset Preparation

import pandas as pd

ml_features = [
    "heart_rate",
    "systolic_bp",
    "diastolic_bp",
    "blood_glucose",
    "body_temperature",
    "sleep_hours",
    "mood_score",
    "activity_minutes",
    "spo2",
    "hr_rolling_3d",
    "sleep_rolling_3d",
    "systolic_rolling_3d",
    "mood_rolling_3d",
    "hr_change",
    "sleep_change",
    "hr_deviation",
    "sleep_deviation",
    "systolic_deviation",
    "mood_deviation"
]

# Keep rows with complete engineered features
ml_df = df.dropna(subset=ml_features).copy()

print("ML Dataset Shape:", ml_df.shape)
print("Remaining Missing Values:",
      ml_df[ml_features].isnull().sum().sum())

# Save full dataset with IDs and timestamps
ml_df.to_csv(
    "astronaut_ml_ready.csv",
    index=False
)

print("ML dataset saved successfully!")
```

```python

# Step 6.2 — Raw Data for Kafka

raw_columns = [
    "reading_id",
    "astronaut_id",
    "timestamp",
    "mission_day",
    "heart_rate",
    "systolic_bp",
    "diastolic_bp",
    "blood_glucose",
    "glucose_context",
    "body_temperature",
    "sleep_hours",
    "mood_score",
    "activity_minutes",
    "spo2",
    "cabin_co2_ppm",
    "cabin_temp_c",
    "cabin_humidity_pct",
    "radiation_dose_rate_uSv_h",
    "data_source"
]

raw_df = df[raw_columns].copy()

# Ensure consistent timestamp formatting
raw_df["timestamp"] = (
    pd.to_datetime(raw_df["timestamp"], utc=True)
      .dt.strftime("%Y-%m-%dT%H:%M:%SZ")
)

raw_df.to_csv(
    "astronaut_kafka_raw.csv",
    index=False
)

print("Kafka Dataset Shape:", raw_df.shape)
print("Kafka raw dataset saved successfully!")
```

```python

import pandas as pd
import zipfile
from pathlib import Path

# 1. Data Dictionary
descriptions = {
    "reading_id": ("Unique reading ID", "ID"),
    "astronaut_id": ("Astronaut identifier", "ID"),
    "timestamp": ("Measurement time in UTC", "ISO 8601"),
    "mission_day": ("Day of mission", "day"),
    "heart_rate": ("Heart rate", "bpm"),
    "systolic_bp": ("Systolic blood pressure", "mmHg"),
    "diastolic_bp": ("Diastolic blood pressure", "mmHg"),
    "blood_glucose": ("Blood glucose", "mg/dL"),
    "glucose_context": ("Glucose measurement context", "category"),
    "body_temperature": ("Body temperature", "Celsius"),
    "sleep_hours": ("Daily sleep duration", "hours"),
    "mood_score": ("Self-reported mood score", "1-5"),
    "activity_minutes": ("Daily activity", "minutes"),
    "spo2": ("Oxygen saturation", "%"),
    "cabin_co2_ppm": ("Cabin CO2 concentration", "ppm"),
    "cabin_temp_c": ("Cabin temperature", "Celsius"),
    "cabin_humidity_pct": ("Cabin relative humidity", "%"),
    "radiation_dose_rate_uSv_h": ("Radiation dose rate", "uSv/h"),
    "data_source": ("Origin of data", "category")
}

feature_descriptions = {
    "hr_rolling_3d": ("Previous 3-day HR average", "bpm"),
    "sleep_rolling_3d": ("Previous 3-day sleep average", "hours"),
    "systolic_rolling_3d": ("Previous 3-day systolic BP average", "mmHg"),
    "mood_rolling_3d": ("Previous 3-day mood average", "score"),
    "hr_change": ("HR minus previous reading", "bpm"),
    "sleep_change": ("Sleep minus previous reading", "hours"),
    "hr_deviation": ("HR minus rolling baseline", "bpm"),
    "sleep_deviation": ("Sleep minus rolling baseline", "hours"),
    "systolic_deviation": ("BP minus rolling baseline", "mmHg"),
    "mood_deviation": ("Mood minus rolling baseline", "score")
}

descriptions.update(feature_descriptions)

dictionary = pd.DataFrame([
    {
        "column": col,
        "dtype": str(ml[col].dtype),
        "description": descriptions.get(col, ""),
        "unit": descriptions.get(col, ("", ""))[1]
    }
    for col in ml.columns
])

dictionary.to_csv("data_dictionary.csv", index=False)

# 2. README
readme = """# Astronaut Health Monitoring - Data Package

## Data Source
Synthetic demonstration data.
NOT real NASA astronaut medical records.

## Files
- astronaut_kafka_raw.csv: 180 records, 19 columns.
- astronaut_ml_ready.csv: 162 records, 29 columns.
- data_preprocessing.ipynb: Data analysis and processing.
- data_dictionary.csv: Column definitions and units.

## Processing
- Missing-value and duplicate checks
- Technical range validation
- Timestamp conversion to UTC
- EDA and correlation analysis
- Astronaut-specific rolling features
- Personal baseline deviations

## Important
The first 3 days per astronaut have insufficient
history for 3-day rolling features. These records
remain in the raw dataset but are excluded from
the ML-ready dataset.

Rolling features use previous observations only.

## For Kafka Team
Send raw readings in chronological order.
Do not send precomputed rolling features as
if they were raw sensor measurements.

## For ML Team
Use a chronological train/test split.
Do not treat synthetic patterns as clinical truth.
Anomaly detection is not medical diagnosis.
"""

Path("README.md").write_text(readme, encoding="utf-8")

# 3. Package
files_to_package = [
    "astronaut_kafka_raw.csv",
    "astronaut_ml_ready.csv",
    "data_dictionary.csv",
    "README.md"
]

if Path("data_preprocessing.ipynb").exists():
    files_to_package.append("data_preprocessing.ipynb")

with zipfile.ZipFile("data_handoff.zip", "w") as z:
    for filename in files_to_package:
        z.write(filename)

print("DATA HANDOFF PACKAGE CREATED")
print(files_to_package)
```
